### What changes from B2 to C1 in speaking

| B2 | C1 |
|---|---|
| Can keep going, but with noticeable pauses to search for words | Speaks **fluently and spontaneously**; rarely has to search for expressions |
| Uses a limited set of linkers (*and, but, because, so*) | Uses a **wide range of connectors and discourse markers** |
| States opinions directly | **Nuances** opinions: hedging, softening, conceding |
| Gets stuck when a word is missing | **Rephrases** smoothly (*"it's a kind of..."*, *"what I mean is..."*) |
| Pronunciation clear but with strong L1 features | Can **vary intonation and stress** to express finer shades of meaning |

You can't become fluent by studying speaking - you need **hours of speaking**. This notebook gives you:
1. functional language to make your speech sound C1,
2. pronunciation work targeted at Spanish speakers,
3. a system for **daily timed speaking practice** with recording and self-assessment.

## 1. Functional language bank

Learn these as **chunks** - say them aloud until they come out automatically.

**Giving opinions with nuance**
*As I see it, ... / If you ask me, ... / I'd argue that... / My take on it is that... / I'm inclined to think that... /
I tend to feel that... / I'm fairly convinced that... / For me, the key issue is...*

**Hedging and softening** (very important in English-speaking workplaces)
*It could be argued that... / To some extent... / There's a case for... / I'm not entirely sure that... /
It might be worth considering... / That's not necessarily the case. / It seems to me that... / I may be wrong, but...*

**Agreeing and disagreeing diplomatically**
*I couldn't agree more. / That's a fair point. / I see where you're coming from, but... / I take your point, but... /
I'm not sure I'd go that far. / I'd see it slightly differently. / With respect, I think... / That's true up to a point.*

**Speculating**
*I'd imagine that... / Chances are... / It's bound to... / I wouldn't be surprised if... / There's a good chance that... /
It's highly unlikely that... / In all likelihood...*

**Buying time** (natural fillers instead of silence or "eh...")
*That's a good question. / Let me think... / How can I put this? / Well, it depends... / I've never really thought about it, but... /
The thing is, ... / What I'm trying to say is...*

**Rephrasing when you don't know a word**
*It's a kind of... / It's the thing you use to... / What's the word... / Let me put it another way. / In other words, ... /
What I mean is...*

**Structuring a longer turn**
*There are a couple of reasons for this. First of all, ... / On top of that, ... / Another thing is... / That said, ... /
On the other hand, ... / All in all, ... / So, to sum up, ... / Coming back to your question, ...*

**Emphasising**
*What really matters is... / The point is... / The last thing we need is... / I do think that... / By far the biggest problem is...*

## 2. Pronunciation for Spanish speakers

Focus on **intelligibility and naturalness**, not a "perfect" accent. These are the features that affect Spanish speakers most:

| Issue | Spanish tendency | Practise |
|---|---|---|
| **Vowel length / quality** | 5 vowels vs ~12 in English | /iː/ vs /ɪ/: *sheep/ship, leave/live, feel/fill, bean/bin* |
| **Schwa /ə/** | pronouncing every vowel fully | the most common sound in English: *a*bout, doct*o*r, *to*day, comf*or*t*a*ble |
| **Initial s + consonant** | adding "e": *"espeak", "estudent"* | *speak, student, Spain, school, strong* - start with a hissing /s/ |
| **Final consonants & clusters** | dropping them | *asked* /ɑːskt/, *months*, *texts*, *world*, *twelfth* |
| **-ed endings** | always /ed/ | /t/ *worked*, /d/ *played*, /ɪd/ only after t/d: *wanted, needed* |
| **/b/ vs /v/** | same sound | *berry / very, ban / van, best / vest* - /v/ is teeth on the lower lip |
| **/θ/ /ð/** | /t/, /d/, /s/ | *think, three, through / this, the, weather* - tongue between the teeth |
| **/h/** | Spanish *j* /x/ (too strong) | *house, hotel, behind* - just a soft breath |
| **/dʒ/ vs /j/** | mixing | *jet / yet, jail / Yale, juice / use* |
| **Word stress** | wrong syllable | *deVELop, PHOtograph, phoTOgrapher, photoGRAphic, eCOnomy, ecoNOmic* |
| **Sentence stress and rhythm** | syllable-timed (every syllable equal) | English is **stress-timed**: stress content words, reduce the rest: *I WANT to GO to the SHOP* |

## 3. Shadowing (15 minutes a day)

Shadowing = speaking **at the same time** as a native speaker, copying their rhythm, stress, intonation and linking.
It trains your mouth and your ear together.

1. Choose a 1-2 minute clip **with a transcript** (TED talk, podcast with transcript, a dictation from Notebook 06).
2. Listen once while reading.
3. Read aloud *with* the audio, a split second behind. Pause and repeat difficult phrases.
4. Shadow **without** the transcript.
5. Record your final version and compare it with the original. What's different? Stress? Speed? Linking?

Repeat the **same** clip for 3-5 days before moving on. Depth beats variety here.

## 5. Real conversations (the most important part)

Grammar and vocabulary only become fluency through **real interaction**. Aim for **at least 2 conversations a week**:

* **Paid tutors / conversation partners:** italki, Preply, Cambly - ask for "conversation practice at C1 level with error correction at the end".
* **Free exchanges:** Tandem, HelloTalk, local language exchange meetups (Meetup.com), Couchsurfing hangouts.
* **Public speaking:** Toastmasters International clubs (many in Spanish-speaking cities run in English) - superb for C1 fluency.
* **AI conversation partner** (for daily low-pressure practice, e.g. in a voice mode). Paste this prompt:

```
You are my English conversation partner. My level is B2 and I'm working towards C1.
Talk with me about [TOPIC] in a natural, conversational way. Ask follow-up questions and
challenge my opinions a little. Don't correct me during the conversation. When I type
"FEEDBACK", give me: (1) my 5 most important mistakes with corrections, (2) 5 more natural
or more advanced ways to say things I said, (3) 3 C1 collocations or idioms I could have used.
```

**After every conversation**, write down 3 things you wanted to say but couldn't, find out how to say them,
and add them to your flashcards. This "gap noticing" is one of the most effective habits you can build.
