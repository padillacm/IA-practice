A **collocation** is a combination of words that native speakers naturally use together:
*heavy rain* (not *strong rain*), *make a decision*, *deeply concerned*, *a vast majority*.

Why this matters so much for you: most B2 speakers already know enough *words*. What separates B2 from C1 is
knowing **which words go together**. Wrong collocations don't block communication, but they instantly signal "non-native"
and they slow listeners down. Research on learner English (e.g. by Michael Lewis, *The Lexical Approach*) shows that
fluent speakers rely on thousands of pre-assembled "chunks" - that's also what makes them fast.

### The main types

| Type | Examples |
|---|---|
| verb + noun | *make* a mistake, *take* a risk, *pay* attention, *raise* a question, *meet* a deadline |
| adjective + noun | *heavy* traffic, *strong* coffee, *bitter* disappointment, *keen* interest, *vast* majority |
| adverb + adjective | *highly* unlikely, *deeply* concerned, *bitterly* disappointed, *well* aware, *fully* booked |
| verb + adverb | *vary considerably*, *rise sharply*, *strongly* recommend, *firmly* believe |
| noun + of + noun | *a stroke of luck*, *a sense of humour*, *a matter of time*, *a flood of complaints* |
| verb + preposition | *depend on*, *consist of*, *object to*, *result in / from*, *account for* |

### How to learn collocations (the method)
1. **Never learn a word alone.** Learn it with its partners: not *decision* but *make / reach / take a decision; a tough decision*.
2. **Use a collocation dictionary**: Oxford Collocations Dictionary (free online at ozdic.com) or SkELL (skell.sketchengine.eu).
3. **Notice** them when you read and listen. Highlight chunks, not single words.
4. **Put them in your SRS deck** (Notebook 08) - there is a ready-made `collocations` deck.

## 1. Verb + noun: the "delexical" verbs

*Make, do, take, have, give, pay, keep, raise, reach, meet, draw* carry little meaning on their own - the noun decides.

| Verb | Typical partners |
|---|---|
| **make** | a decision, a mistake, progress, an effort, a complaint, a difference, an impression, a point, sense, an exception, arrangements |
| **do** | business, damage, harm, research, a favour, your best, the shopping, a course, justice to |
| **take** | a risk, a break, action, steps, place, part in, advantage of, responsibility for, into account, care of |
| **pay** | attention, a visit, a compliment, tribute to, the price |
| **raise** | a question, a concern, awareness, money, the issue, expectations |
| **reach / come to** | an agreement, a decision, a conclusion, a compromise, a consensus |
| **meet** | a deadline, a need, expectations, requirements, demand |
| **draw** | a conclusion, attention to, a distinction, a comparison, inspiration from |
| **keep** | an eye on, a promise, a secret, in touch, track of |
| **run** | a risk, a business, late, out of |

## 2. Adjective + noun and adverb + adjective

Using **precise, strong collocations instead of *very* + adjective** is one of the fastest ways to sound C1.

| Instead of... | Say... |
|---|---|
| very big problem | a **major / serious / pressing** problem |
| very important | **crucial, vital, essential, of paramount importance** |
| a lot of rain | **heavy / torrential** rain |
| very sad / disappointed | **deeply** saddened, **bitterly** disappointed |
| very sure | **absolutely** certain, **firmly** convinced |
| very different | **strikingly / fundamentally / radically** different |
| very unlikely | **highly** unlikely |
| very aware | **well / fully / acutely** aware |
| very hard work | **gruelling / demanding / painstaking** work |
| very good reason | a **compelling / valid** reason |

## 3. Collocations for discussing ideas, data and change

These are the collocations you need in meetings, presentations, discussions and the news.

**Arguments and evidence:** raise / make a **point**, a **valid** point, **shed / throw** light on, **cast doubt** on,
**hold water**, **strike a balance**, **take into account**, **far-reaching** consequences, **broad / widespread** consensus,
**fierce / strong** opposition, **conflicting** evidence, **draw** a distinction.

**Trends:** rise / fall **sharply / dramatically / steadily / slightly**; a **steady / gradual / sharp** decline;
**reach a peak**, **hit a low**, **level off**, **bridge / close / narrow the gap**.
